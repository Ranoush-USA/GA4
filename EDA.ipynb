```python
%sql
SELECT event_date  ,event_name,count(*)as row_count 

FROM workspace.bronze.events
where event_name='session_start'
group by  event_date,event_name
order by row_count desc
```

```python
%sql
-- Check for any remaining nulls
SELECT
    COUNT(*) AS total,
    COUNT(user_id) AS has_user_id,
    COUNT(event_timestamp) AS has_timestamp,
    COUNT(page_location) AS has_page_location,
    COUNT(page_title) AS has_page_title
FROM workspace.silver.ga4_silver;
```

```python
%sql
-- Check date range
SELECT
    MIN(event_timestamp) AS earliest,
    MAX(event_timestamp) AS latest
FROM workspace.silver.ga4_silver;
```

```python
%sql

-- Top events
SELECT event_name, COUNT(*) as count
FROM workspace.silver.ga4_silver
GROUP BY event_name
ORDER BY count DESC;
```
