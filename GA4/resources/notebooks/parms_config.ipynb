### Why using widgets for Params here: 

The databricks.yml identify the default values for paramaters, howerver,the notebook widget default servers two purposes here:

- A parameter the job never passes (e.g. gold_schema in the Silver task, bronze_schema in the Gold task) — permanently falls back to the 
notebook's hardcoded default.

- Running the notebook manually/interactively, outside the job (e.g. testing in the workspace UI) — there's no job passing parameters at all, so the widget default is what you get.

When the job does pass the parameter, the notebook default is simply overridden and never used — so it is, "safety net"

```python
# Create widgets with default values
dbutils.widgets.text("catalog", "workspace", "Catalog")

dbutils.widgets.text("silver_schema", "silver", "Silver_Schema")
dbutils.widgets.text("bronze_schema", "bronze", "Bronze_Schema")
dbutils.widgets.text("gold_schema", "gold")


dbutils.widgets.text("bronze_table", "events")
dbutils.widgets.text("silver_table", "ga4_silver")
# Then retrieve the values
catalog = dbutils.widgets.get("catalog")

bronze_schema = dbutils.widgets.get("bronze_schema")
silver_schema = dbutils.widgets.get("silver_schema")
gold_schema = dbutils.widgets.get("gold_schema")

bronze_table = dbutils.widgets.get("bronze_table")
silver_table = dbutils.widgets.get("silver_table")


# Create the schema if not exist 
spark.sql(f"USE {catalog}.{bronze_schema}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{silver_schema}")
```
